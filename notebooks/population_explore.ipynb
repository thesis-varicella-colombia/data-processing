### Read excel files

```python

from pathlib import Path
import pandas as pd

data_dir = Path("data/raw/population")
if not data_dir.is_dir():
    data_dir = Path("../data/raw/population")

pop_2005_2017 = pd.read_excel(
    data_dir / "DCD-area-sexo-edad-proypoblacion-Mun-2005-2017_VP.xlsx",
    sheet_name="NuevaMpal",
    header=11,
    dtype={"MPIO": str},
)

pop_2018_2042 = pd.read_excel(
    data_dir / "PPED-AreaSexoEdadMun-2018-2042_VP.xlsx",
    sheet_name="PobMunicipalxÁreaSexoEdad",
    header=8,
    dtype={"Unnamed: 2": str},
)

# Los nombres de las primeras seis columnas están en la fila anterior.
pop_2018_2042.columns = [
    "DP", "DPNOM", "MPIO", "DPMP", "AÑO", "ÁREA GEOGRÁFICA",
    *pop_2018_2042.columns[6:],
]

print("2005–2017:", pop_2005_2017.shape)
print("2018–2042:", pop_2018_2042.shape)
```

```python
print(", ".join(pop_2005_2017.columns))
```

```python
print(", ".join(pop_2018_2042.columns))
```

```python

grupos_edad = {
    "<1": [0],
    "1-4": range(1, 5),
    "5-14": range(5, 15),
    "15-49": range(15, 50),
    "50+": range(50, 85),
}

pop_2005_2017_grupos = pop_2005_2017[
    ["DP", "DPNOM", "DPMP", "MPIO", "AÑO", "ÁREA GEOGRÁFICA"]
].copy()

for sexo in ("Hombres", "Mujeres", "Total"):
    for grupo, edades in grupos_edad.items():
        columnas = [f"{sexo}_{edad}" for edad in edades]
        if grupo == "50+":
            columnas.append(f"{sexo}_85 y más")

        pop_2005_2017_grupos[f"{sexo}_{grupo}"] = (
            pop_2005_2017[columnas].sum(axis=1, min_count=1)
        )

display(pop_2005_2017_grupos.head())
```

```python
pop_2005_2017_grupos.columns
```

```python
pop_2018_2042_grupos = pop_2018_2042[
    ["DP", "DPNOM", "MPIO", "DPMP", "AÑO", "ÁREA GEOGRÁFICA"]
].copy()

for sexo in ("Hombres", "Mujeres", "Total"):
    for grupo, edades in grupos_edad.items():
        columnas = [
            f"{sexo} {edad} {'año' if edad == 1 else 'años'}"
            for edad in edades
        ]

        if grupo == "50+":
            columna_100_mas = next(
                col for col in pop_2018_2042.columns
                if col.startswith(f"{sexo} 100 años") and col.endswith("y más")
            )
            columnas.append(columna_100_mas)

        pop_2018_2042_grupos[f"{sexo}_{grupo}"] = (
            pop_2018_2042[columnas].sum(axis=1, min_count=1)
        )

display(pop_2018_2042_grupos.head())
```

```python
pop_2018_2042_grupos.columns
```

```python
poblacion_grupos = pd.concat(
    [pop_2005_2017_grupos, pop_2018_2042_grupos],
    ignore_index=True,
)

# Conservar los códigos municipales de cinco dígitos
poblacion_grupos["MPIO"] = poblacion_grupos["MPIO"].astype("string").str.zfill(5)

print(poblacion_grupos.shape)
display(poblacion_grupos.head())
```

```python
poblacion_grupos = poblacion_grupos[
    poblacion_grupos["AÑO"].between(2007, 2025, inclusive="both")
].copy()

# Mantener códigos con ceros iniciales y conteos enteros al exportar.
poblacion_grupos["DP"] = poblacion_grupos["DP"].astype("string").str.zfill(2)
poblacion_grupos["AÑO"] = poblacion_grupos["AÑO"].astype(int)
columnas_grupos = [col for col in poblacion_grupos if col.startswith(("Hombres_", "Mujeres_", "Total_"))]
poblacion_grupos[columnas_grupos] = poblacion_grupos[columnas_grupos].astype(int)

print(poblacion_grupos["AÑO"].min(), poblacion_grupos["AÑO"].max())
```

```python
with pd.option_context(
        "display.max_rows", None,
        "display.max_columns", None,
        "display.max_colwidth", None,
):
    display(poblacion_grupos.head(20))
```

```python
poblacion_grupos["AÑO"] = pd.to_numeric(
    poblacion_grupos["AÑO"], errors="coerce"
).astype("Int64")

for nombre in poblacion_grupos.columns:
    if nombre.startswith(("Hombres_", "Mujeres_", "Total_")):
        poblacion_grupos[nombre] = pd.to_numeric(
            poblacion_grupos[nombre],
            errors="coerce"
        ).astype("Int64")

poblacion_grupos.dtypes
```

```python
# usar los grupos de edad recomendado por el INS para el seguimiento de la varicella
poblacion_clean = poblacion_grupos.copy()

poblacion_clean["codigo_municipio"] = (
    poblacion_clean["MPIO"]
    .astype(str)
    .str.strip()
    .str.zfill(5)
)

poblacion_clean["ano_epidemiologico"] = pd.to_numeric(
    poblacion_clean["AÑO"], errors="coerce"
).astype("Int64")

poblacion_clean["poblacion_menor_1"] = pd.to_numeric(
    poblacion_clean["Total_<1"], errors="coerce"
).astype("Int64")

poblacion_clean["poblacion_1_4"] = pd.to_numeric(
    poblacion_clean["Total_1-4"], errors="coerce"
).astype("Int64")

poblacion_clean["poblacion_5_14"] = pd.to_numeric(
    poblacion_clean["Total_5-14"], errors="coerce"
).astype("Int64")

poblacion_clean["poblacion_15_49"] = pd.to_numeric(
    poblacion_clean["Total_15-49"], errors="coerce"
).astype("Int64")

poblacion_clean["poblacion_50_mas"] = pd.to_numeric(
    poblacion_clean["Total_50+"], errors="coerce"
).astype("Int64")

columnas_poblacion = [
    "poblacion_menor_1",
    "poblacion_1_4",
    "poblacion_5_14",
    "poblacion_15_49",
    "poblacion_50_mas",
]

poblacion_clean["poblacion_total"] = (
    poblacion_clean[columnas_poblacion]
    .fillna(0)
    .sum(axis=1)
    .astype("Int64")
)

poblacion_clean.head()
```

```python
df = poblacion_clean.copy()

# Claves
df["codigo_municipio"] = (
    df["MPIO"]
    .astype(str)
    .str.strip()
    .str.zfill(5)
)

df["ano_epidemiologico"] = pd.to_numeric(
    df["AÑO"], errors="coerce"
).astype("Int64")

# Columnas originales de población
cols_total = [
    "Total_<1",
    "Total_1-4",
    "Total_5-14",
    "Total_15-49",
    "Total_50+",
]

cols_hombres = [
    "Hombres_<1",
    "Hombres_1-4",
    "Hombres_5-14",
    "Hombres_15-49",
    "Hombres_50+",
]

cols_mujeres = [
    "Mujeres_<1",
    "Mujeres_1-4",
    "Mujeres_5-14",
    "Mujeres_15-49",
    "Mujeres_50+",
]

# Equivalente a try_cast
for c in cols_total + cols_hombres + cols_mujeres:
    df[c] = pd.to_numeric(df[c], errors="coerce")

# Totales por fila
# min_count=5 se parece más al comportamiento de Spark:
# si alguno de los componentes es null, la suma queda null.
df["_total"] = df[cols_total].sum(axis=1, min_count=5)
df["_hombres"] = df[cols_hombres].sum(axis=1, min_count=5)
df["_mujeres"] = df[cols_mujeres].sum(axis=1, min_count=5)

# Crear las columnas que después serán agregadas con max()
es_total = df["ÁREA GEOGRÁFICA"].eq("Total")
es_cabecera = df["ÁREA GEOGRÁFICA"].eq("Cabecera Municipal")
es_rural = df["ÁREA GEOGRÁFICA"].eq(
    "Centros Poblados y Rural Disperso"
)

df["poblacion_total"] = df["_total"].where(es_total)
df["poblacion_hombres"] = df["_hombres"].where(es_total)
df["poblacion_mujeres"] = df["_mujeres"].where(es_total)

df["poblacion_cabecera"] = df["_total"].where(es_cabecera)
df["poblacion_rural"] = df["_total"].where(es_rural)

# Grupos de edad - total
df["poblacion_menor_1"] = df["Total_<1"].where(es_total)
df["poblacion_1_4"] = df["Total_1-4"].where(es_total)
df["poblacion_5_14"] = df["Total_5-14"].where(es_total)
df["poblacion_15_49"] = df["Total_15-49"].where(es_total)
df["poblacion_50_mas"] = df["Total_50+"].where(es_total)

# Hombres por grupo de edad
df["poblacion_hombres_menor_1"] = df["Hombres_<1"].where(es_total)
df["poblacion_hombres_1_4"] = df["Hombres_1-4"].where(es_total)
df["poblacion_hombres_5_14"] = df["Hombres_5-14"].where(es_total)
df["poblacion_hombres_15_49"] = df["Hombres_15-49"].where(es_total)
df["poblacion_hombres_50_mas"] = df["Hombres_50+"].where(es_total)

# Mujeres por grupo de edad
df["poblacion_mujeres_menor_1"] = df["Mujeres_<1"].where(es_total)
df["poblacion_mujeres_1_4"] = df["Mujeres_1-4"].where(es_total)
df["poblacion_mujeres_5_14"] = df["Mujeres_5-14"].where(es_total)
df["poblacion_mujeres_15_49"] = df["Mujeres_15-49"].where(es_total)
df["poblacion_mujeres_50_mas"] = df["Mujeres_50+"].where(es_total)

# Columnas finales a agregar
columnas_agregadas = [
    "poblacion_total",
    "poblacion_hombres",
    "poblacion_mujeres",
    "poblacion_cabecera",
    "poblacion_rural",
    "poblacion_menor_1",
    "poblacion_1_4",
    "poblacion_5_14",
    "poblacion_15_49",
    "poblacion_50_mas",
    "poblacion_hombres_menor_1",
    "poblacion_hombres_1_4",
    "poblacion_hombres_5_14",
    "poblacion_hombres_15_49",
    "poblacion_hombres_50_mas",
    "poblacion_mujeres_menor_1",
    "poblacion_mujeres_1_4",
    "poblacion_mujeres_5_14",
    "poblacion_mujeres_15_49",
    "poblacion_mujeres_50_mas",
]

poblacion_to_join = (
    df
    .groupby(
        ["codigo_municipio", "ano_epidemiologico"],
        as_index=False
    )[columnas_agregadas]
    .max()
)

poblacion_to_join[columnas_agregadas] = (
    poblacion_to_join[columnas_agregadas].astype("Int64")
)

poblacion_to_join.head()
```

```python
# Guardar la tabla final por municipio y año.
output_dir = data_dir.parent.parent / "processed/population"
output_dir.mkdir(parents=True, exist_ok=True)
output_file = output_dir / "population_municipal_2007_2025.csv"
poblacion_to_join.to_csv(output_file, index=False, encoding="utf-8")
print(f"Guardado: {output_file} ({len(poblacion_to_join):,} filas)")
```
